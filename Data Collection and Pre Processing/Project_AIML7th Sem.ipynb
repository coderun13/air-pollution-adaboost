```python
# Data Collection and Preprocessing
# Air Pollution Dataset: India 2017-2025

import pandas as pd
import numpy as np

# Optional libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
# Load the dataset

file_path = "IndiaAirPollution_2017_25.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)
```

```python
# Display all column names

for i, column in enumerate(df.columns, start=1):
    print(i, column)
```

```python
# Dataset information

df.info()
```

```python
# Statistical summary of categorical columns

df.describe(include='object')
```

```python
# Count missing values in every column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Show only columns having missing values

missing = df.isnull().sum()
missing = missing[missing > 0]

if len(missing) == 0:
    print("No missing values found in the dataset.")
else:
    print(missing)
```

```python
# Calculate percentage of missing values

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_table = missing_table[missing_table['Missing Count'] > 0]

if missing_table.empty:
    print("No missing values found.")
else:
    display(missing_table)
```

```python
# Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Convert Datetime column to datetime type

df['Datetime'] = pd.to_datetime(df['Datetime'], errors='coerce')

print(df['Datetime'].dtype)
```

```python
# Check invalid datetime values

invalid_dates = df['Datetime'].isnull().sum()

print("Invalid datetime values:", invalid_dates)
```

```python
# Sort records chronologically

df = df.sort_values('Datetime').reset_index(drop=True)

df.head()
```

```python
# Identify numerical columns

numeric_columns = df.select_dtypes(include=[np.number]).columns

print("Number of numerical columns:", len(numeric_columns))
```

```python
# Fill missing numerical values with median

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values handled.")
```

```python
# Identify categorical columns

categorical_columns = df.select_dtypes(include=['object']).columns

print("Categorical columns:")
print(categorical_columns.tolist())

# Fill missing categorical values using mode

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("Missing categorical values handled.")
```

```python
# Check remaining missing values

print("Total missing values:", df.isnull().sum().sum())
```

```python
# Remove duplicate rows

print("Duplicate rows before removal:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after removal:", df.duplicated().sum())
```

```python
# Convert Datetime column

if 'Datetime' in df.columns:
    df['Datetime'] = pd.to_datetime(df['Datetime'], errors='coerce')
    df = df.sort_values('Datetime').reset_index(drop=True)

    print("Datetime converted and data sorted successfully.")
else:
    print("Datetime column not found.")
```

```python
# Major pollution columns

pollution_columns = [
    'PM2.5', 'PM10', 'NO', 'NO2', 'NOx',
    'NH3', 'CO', 'SO2', 'O3',
    'Benzene', 'Toluene', 'Xylene'
]

# Use only columns present in dataset
pollution_columns = [
    col for col in pollution_columns if col in df.columns
]

print("Pollution columns:", pollution_columns)

# Replace negative values with NaN
for column in pollution_columns:
    df.loc[df[column] < 0, column] = np.nan

# Fill NaN with median
for column in pollution_columns:
    df[column] = df[column].fillna(df[column].median())

print("Negative pollution values handled.")
```

```python
# PM2.5 distribution

plt.figure(figsize=(10, 5))

plt.hist(df['PM2.5'], bins=50)

plt.title('Distribution of PM2.5')
plt.xlabel('PM2.5 Concentration')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap

numeric_data = df.select_dtypes(include=[np.number])

plt.figure(figsize=(16, 10))

sns.heatmap(
    numeric_data.corr(),
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Heatmap of Numerical Features')
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px

# Interactive PM2.5 graph

fig = px.line(
    df,
    x='Datetime',
    y='PM2.5',
    title='PM2.5 Levels Over Time',
    labels={
        'Datetime': 'Date',
        'PM2.5': 'PM2.5 Concentration'
    }
)

fig.show()
```

```python
# Interactive AQI graph

if 'AQI' in df.columns:

    fig = px.line(
        df,
        x='Datetime',
        y='AQI',
        title='Air Quality Index (AQI) Over Time',
        labels={
            'Datetime': 'Date',
            'AQI': 'Air Quality Index'
        }
    )

    fig.show()

else:
    print("AQI column not found.")
```

```python
# Average AQI by city

if 'City' in df.columns and 'AQI' in df.columns:

    city_aqi = (
        df.groupby('City')['AQI']
        .mean()
        .sort_values(ascending=False)
        .reset_index()
    )

    city_aqi.columns = ['City', 'Average_AQI']

    display(city_aqi.head(20))
```

```python
# Interactive city-wise AQI graph

if 'City' in df.columns and 'AQI' in df.columns:

    fig = px.bar(
        city_aqi,
        x='City',
        y='Average_AQI',
        title='Average AQI by City',
        labels={
            'City': 'City',
            'Average_AQI': 'Average AQI'
        }
    )

    fig.update_layout(
        xaxis_tickangle=-45
    )

    fig.show()
```

```python
# Average pollution levels

pollution_average = df[pollution_columns].mean().sort_values(
    ascending=False
)

display(pollution_average)
```

```python
# Interactive pollutant comparison

pollution_df = pollution_average.reset_index()

pollution_df.columns = ['Pollutant', 'Average_Value']

fig = px.bar(
    pollution_df,
    x='Pollutant',
    y='Average_Value',
    title='Average Concentration of Air Pollutants',
    labels={
        'Pollutant': 'Pollutant',
        'Average_Value': 'Average Concentration'
    }
)

fig.show()
```

```python
# Min-Max normalization

df_normalized = df.copy()

numeric_columns = df_normalized.select_dtypes(
    include=[np.number]
).columns

for column in numeric_columns:

    minimum = df_normalized[column].min()
    maximum = df_normalized[column].max()

    if maximum != minimum:
        df_normalized[column] = (
            df_normalized[column] - minimum
        ) / (maximum - minimum)
    else:
        df_normalized[column] = 0

print("Normalization completed successfully.")
```

```python
# Encode categorical columns

categorical_columns = df_normalized.select_dtypes(
    include=['object']
).columns

for column in categorical_columns:
    df_normalized[column] = (
        df_normalized[column]
        .astype('category')
        .cat.codes
    )

print("Categorical encoding completed.")
```

```python
# Final dataset check

print("======================================")
print("FINAL PREPROCESSING RESULT")
print("======================================")

print("Rows:", df_normalized.shape[0])
print("Columns:", df_normalized.shape[1])
print("Missing values:", df_normalized.isnull().sum().sum())
print("Duplicate rows:", df_normalized.duplicated().sum())

print("======================================")
```

```python
# Save preprocessed dataset

df_normalized.to_csv(
    "IndiaAirPollution_Preprocessed.csv",
    index=False
)

print("Final preprocessed dataset saved as:")
print("IndiaAirPollution_Preprocessed.csv")
```
